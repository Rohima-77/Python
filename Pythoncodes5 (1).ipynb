```python
#Association (Simple connection)

#দুইটা class একে অন্যকে “জানে”
#কিন্তু কেউ কারো owner না
#সবচেয়ে weak relationship
# Assosiation

class Doctor:
    def __init__(self, name):
        self.name = name

class Patient:
    def __init__(self, name, doctor):
        self.name = name
        self.doctor = doctor

d1 = Doctor("Rahim")
p1 = Patient("Karim", d1)

print(p1.doctor.name)
```

```python
# Aggregation (Has-A but weak ownership)

# “Has-A relationship”
# parent object থাকলেও child আলাদা থাকতে পারে
# medium strength relationship
```

```python
class Teacher:
    def __init__(self, name):
        self.name = name

class University:
    def __init__(self, name, teacher):
        self.name = name
        self.teacher = teacher

t1 = Teacher("Sakib")

u1 = University("BUBT", t1)

print(u1.teacher.name)# teacher survives without university so it is Aggregation
```

```python
#Composition (Strong ownership)

#Strong “Part-of” relationship
#Parent ছাড়া child বাঁচে না
#সবচেয়ে strong relationship
```

```python
# Composition
class Room:
    def show(self):
        return "This is a room"

class House:
    def __init__(self):
        self.room = Room()   # inside create

    def display(self):
        return self.room.show()

h1 = House()

print(h1.display())
```

```python
# Inheritance
#1.Single inheritance(1 parent.........1 child)
#parent class
class Grandfather:
    def __init__(self,color,last_name):
      self.color = color
      self.last_name = last_name
# child class
class Father(Grandfather):
    def __init__(self,name,color,last_name):
      super().__init__(color,last_name)# etar madhome parent class er characteristics gulo child class er modhe access korlam
      self.name = name

gd = Grandfather("Blue","Khan")
f1 = Father("Rahim","Red","Khan")
print(f1.name)
print(f1.color)
print(f1.last_name)
```

```python
# single inheritance practice
class Person:
    def __init__(self,name,age):
      self.name = name
      self.age = age

class Employee(Person):
    def __init__(self,salary,name,age):
      super().__init__(name,age)
      self.salary = salary

p1 = Person("Turi","32")
e1 = Employee("50000","Turi","32")
print(e1.name,e1.age,e1.salary)
```

```python
# more practice in single inheritance
class Vehicle:
    def __init__(self,brand,year):
      self.brand = brand
      self.year = year
class Child(Vehicle):
    def __init__(self,model,brand,year):
      super().__init__(brand,year)
      self.model =model

v1 = Vehicle("Toyota",2022)
c1 = Child("Hon","Toyota",2022)
print(c1.model,c1.brand,c1.year)

```

```python
# Multiple inheritance
#একটা child class একাধিক parent class থেকে feature নেয়।
#Grandfather---Father---child
class Father:
    def __init__(self,first_name):
      self.first_name = first_name
class Mother:
    def __init__(self,behavior):
      self.behavior = behavior
class Child(Father,Mother):
    def __init__(self,name,first_name,behavior):
      Father.__init__(self,first_name)
      Mother.__init__(self,behavior)
      self.name = name
c1 = Child("Rahim","Karim","Good")
print(c1.name,c1.first_name,c1.behavior)
```

```python
#Multilevel inheritance
class Grandfather:
    def __init__(self,f_name):
      self.f_name = f_name

class Father(Grandfather):
    def __init__(self,f_name,behavior):
      self.behavior = behavior
      super().__init__(f_name)


class Child(Father):
    def __init__(self,f_name,behavior,l_name):
      super().__init__(f_name,behavior)
      self.l_name = l_name


c1 = Child("Rahim","Good","Khan")
print(c1.f_name,c1.behavior,c1.l_name)
```

```python
# Hierarchical inheritance
# One parent but Multiple children
# prottekta child parent ke inherit kore
class Person:
    def university(self):
      print("BuBT")
class Student1(Person):
    def attendance(self):
      print("90%")

class Student2(Person):
    def final_marks(self):
      print("Betetr than mid term marks")

s1 = Student1()
s1.university()
s1.attendance()
s2 = Student2()
s2.university()
s2.final_marks()
```

```python
class Person:

    def walk(self):
        print("Person can walk")


class Student(Person):

    def study(self):
        print("Student is studying")


class Teacher(Person):

    def teach(self):
        print("Teacher is teaching")


s1 = Student()
t1 = Teacher()

s1.walk()
s1.study()

t1.walk()
t1.teach()
```

```python
# Hybrid inheritance
# Main parent class
class Device:

    def power_on(self):
        print("Device Power ON")


# Child class 1
class Camera(Device):

    def take_photo(self):
        print("Taking Photo")


# Child class 2
class MusicPlayer(Device):

    def play_music(self):
        print("Playing Music")


# Hybrid class
class SmartPhone(Camera, MusicPlayer):

    def call(self):
        print("Calling Someone")


# object create
s1 = SmartPhone()


# methods call
s1.power_on()
s1.take_photo()
s1.play_music()
s1.call()
```

```python
class Parent:
    def behavior(self):
      print("Good")

class Child1(Parent):
    def study(self):
      print("Better")

class Child2(Parent):
    def play(self):
      print("Best")

class Child3(Child1,Child2):
    def sleep(self):
      print("8 hour")

c1 = Child3()
c1.behavior()
c1.study()
c1.play()
c1.sleep()
```
