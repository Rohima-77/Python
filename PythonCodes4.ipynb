```python
# class vs instance variable
class Student:
  department = "CSE" #class variable
  def __init__(self,name,age):
    self.name = name # instance variable
    self.age = age #instance variable
s1 = Student("Rohima","21")
s2 = Student("Yuku","19")
print(s1.name,s1.age,s1.department)
print(s2.name,s2.age,s2.department)
```

```python
from typing import AsyncGenerator
from os import name
# class method vs instance method
class Student:
  department = "cse"# class variable
  def __init__(self,name,age):
    self.name = name # instance variable
    self.age = age #instance variable
  # Instance method
  def show_name(self):
    print(self.name)

  def show_age(self):
    print(self.age)


# class method
  @classmethod
  def show_department(cls):
    print(cls.department)


s1 = Student("Rohi","21")
s1.show_name()
s1.show_age()

Student.show_department()



```

```python
# sattic method
# independent helper methode
# where there is no use self or cls
class Grade:
  @staticmethod
  def calculate_grade(marks):
    if marks>=90:
      return 'A+'
    else:
      return 'F'
r1 = Grade.calculate_grade(95)
print(r1)
```

```python
# Encapsulation
# using getter and setter method
# Getter use:Value read
# Setter use :Value set
class Bank:
  def __init__(self,balance):
    self. __balance = balance
  def get_balance(self):
    return self.__balance

  def set_balance(self,amount):
    self.__balance = amount
b1 = Bank(1000)
print(b1.get_balance())# read value
b1.set_balance(2000)# set value
print(b1.get_balance())
```

```python
# Student marks system using encapsulation
class Student:
  def __init__(self,marks):
    self.__marks=marks
  def get_marks(self):
    return self.__marks
  def set_marks(self,marks):
    if marks<=0 or marks<=100:
      self.__marks=marks
    else:
      print("Invalid marks")
mar = Student(80)
print(mar.get_marks())
mar.set_marks(90)
print(mar.get_marks())
```
