```python
# map()
nums = [1,2,3,4,5]
square = list(map(lambda x:x*x,nums))
print(square)
```

```python
# find even numbers using filter()
num = [1,2,3,4,5,6,7,8,9]
even = list(filter(lambda x:x%2==0,num))
print(even)

# find even numbers using filter()
num = [1,2,3,4,5,6,7,8,9]
odd = list(filter(lambda x:x%2!=0,num))
print(odd)
```

```python
# sum of the values using reduce()
from functools import reduce
num = [1,2,3,4,5,6,7,8]
sum = reduce(lambda a,b:a+b,num)
print(sum)
```

```python
# OOP
# define class and objects
# init is a constructor and set the object data
# There are three types of constructor
#1.default constructor(No parameter)
#2.parameterized constructor(argumnet bole dite hoy)
#3.Default value constructor

class Car:
  def __init__(self):# default constructor
    self.brand = " "
    self.model = " "

car1 = Car()
car1.brand = "Toyota"
car1.model = "camry"
print(car1.brand,car1.model)

car2 = Car()
car2.brand = "BMW"
car2.model = "x5"
print(car2.brand,car2.model)
```

```python
# Default value constructor
class Person:
  def __init__(self,skill="python",uni="bubt",dept="CSE"):
    self.skill=skill
    self.uni=uni
    self.dept=dept
per1 = Person()
print(per1.skill,per1.uni,per1.dept)
```

```python
# Adding methode inside the class
class Person:
  def __init__(self,skill="python",uni="bubt",dept="CSE"):# use constructor
    self.skill=skill
    self.uni=uni
    self.dept=dept
  def info(self):# use method
    print(self.skill)
    print(self.uni)
    print(self.dept)
per1 = Person()
per1.info()
```
