```python
# break and continue
# print 1 to 10 but when comes to 5 than break the loop

for i in range(1,11):
  if i==5:
    break;
  print(i)

for i in range(1,11):
  if i==5:
    continue;
  print(i)

#print 1 to 20 odd numbers use continue
for i in range(1,21):
  if i%2!=0:
    print(i)
```

```python
# while loop
# print sum of all list elements
a = [2,2,3,4,5,6]
result = 0
i = 0
n = len(a)
while i<n:
  result = result + a[i]
  i = i+1
print(result)
```

```python
# Function
# No input No return
def my_function():# function deifne
  a = 10
  b = 20
  print(a+b)
my_function()# Function call

# No Input but return
def multiply():
  a = 10
  b = 20
  return(a*b)
result = multiply()
print(result)


# Input but no return
def adding(a,b):# argument
  print(a+b)
adding(10,30)# Parameter


# Input and return
def sub(a,b):
  return(a-b)
result=sub(10,2)
print(result)




```

```python
# multiple arguments
def addition(*args):
  print(args)
  return(sum(args))

result = addition(1,2,3,4,5)
print(result)
```
